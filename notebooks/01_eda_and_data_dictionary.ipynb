#### 2.4 Write a 1-paragraph markdown defense explaining why Exposure acts as a rate offset, and why continuous Density and categorical Area cannot both be included in a linear regression model.
Exposure acts as a rate offset because the number of claims depends on how long a policy is active. For example, a policy that is active for one year has more opportunity to generate claims than a policy that is active for only six months, so claim counts should be adjusted based on Exposure. Density and Area should not both be included in the same linear regression model because they represent similar information about population density. Density gives a continuous numerical value, while Area groups population density into categories. Including both variables would add redundant information to the model and could create collinearity problems.

### Part 3: Overriding Automated Rules & Correlation Visuals

#### 3.1 Render the Pearson Correlation Heatmap in Cell 6b and export reports/figures/correlation_heatmap.png

```python
print("==================================================")
print("UNIVARIATE SKEWNESS DIAGNOSTICS")
print("==================================================")
 
numeric_df = df.select_dtypes(include=[np.number])
skewness_series = numeric_df.skew().sort_values(ascending=False)
 
skew_summary = pd.DataFrame({
    "Feature": skewness_series.index,
    "Skewness Coefficient": skewness_series.values,
    "Diagnosis": ["Extremely Right-Skewed" if s > 3.0 else "Moderately Skewed" if abs(s) > 1.0 else "Unskewed/Symmetric" for s in skewness_series.values],
    "Action Required": ["Log/Box-Cox Transform" if abs(s) > 1.0 else "Standard Scaler Sufficient" for s in skewness_series.values]
})
 
display(skew_summary)
```

```python
plt.figure(figsize=(10, 8))

# Compute Pearson correlation matrix across numerical variables
corr = numeric_df.corr()

# Generate a diverging color map (Cool-Warm: Blue=Negative, Red=Positive)
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1.0,
    vmax=1.0,
    linewidths=0.5,
    cbar_kws={"label": "Pearson Correlation Coefficient (r)"}
)

plt.title(
    "Pearson Correlation Heatmap: Numerical Features",
    fontsize=14,
    fontweight="bold",
    pad=12
)

plt.tight_layout()

heatmap_path = os.path.join(
    "..", "reports", "figures", "correlation_heatmap.png"
)

plt.savefig(heatmap_path, dpi=300)
plt.show()

print(f"[+] Saved correlation heatmap graphics artifact to: {heatmap_path}")
```

#### 3.2 Write an executive defense explaining why you overrode automated skewness recommendations for ClaimNb, detailing the mathematical impossibility (log(0) = −∞) and domain rationale.

The automated skewness analysis identifies ClaimNb as extremely right-skewed, with a skewness coefficient of approximately 5.60, and recommends a log transformation. However, I would override this recommendation because ClaimNb is a discrete count variable, not a continuous positive variable. It can also contain zero values, and log(0) is mathematically undefined and approaches -∞. Therefore, ClaimNb should remain in its original count form and be modeled using a Poisson or Negative Binomial model instead of applying a log transformation.

#### 4.2 Write the required 3-bullet interpretation statements (Distribution Shape, Key Metrics/Whiskers, Recommended Analytics Action) below each of the 4 distribution charts and the correlation heatmap.

##### 1. Box Plot: Population Density (People/km2)
- **Distribution Shape:** Density is extremely right-skewed, with a skewness coefficient of approximately 4.65. Most observations are concentrated at lower population densities, while a long tail extends to the right.
- **Key Metrics/Whiskers:** The upper whisker is around 4,000 people/km², while many observations extend far beyond the whisker, with some values above 20,000 people/km². These points are statistical outliers but may represent valid high-density urban areas.
- **Recommended Analytics Action:** Keep the valid high-density observations and apply a log(1 + Density) transformation to reduce the strong right skew before modeling.

##### 2. Histogram & KDE: Driver Age Distribution
- **Distribution Shape:** DrivAge is relatively symmetric and close to a bell-shaped distribution, with a skewness coefficient of approximately 0.44. 
- **Key Metrics/Whiskers:** Driver ages range from 18 to 100, and most drivers are concentrated between roughly 30 and 60 years old.
- **Recommended Analytics Action:** No log transformation or outlier trimming is needed. Standard Z-score scaling can be used before modeling.

##### 3. Box Plot: BonusMalus Risk Index
- **Distribution Shape:** BonusMalus is moderately right-skewed, with a skewness coefficient of approximately 1.73 with most values concentrated at the lower end of the distribution.
- **Key Metrics/Whiskers:** Most scores are concentrated below 100, while values above the upper whisker appear as outliers and extend to approximately 230.
- **Recommended Analytics Action:** These higher values represent valid high-risk drivers rather than data errors, so they should be retained. BonusMalus can be scaled, and a binary indicator such as Is_Surcharge_Driver = BonusMalus > 100 can also be created.

##### 4. Histogram & KDE: Vehicle Age Distribution
- **Distribution Shape:** VehAge is moderately right-skewed, with a skewness coefficient of approximately 1.15. Most vehicles are concentrated at younger ages and a long tail extending to the right.
- **Key Metrics/Whiskers:** Most vehicles are less than about 20 years old, while only a small number of observations extend to much older vehicle ages.
- **Recommended Analytics Action:** VehAge can be grouped into age categories or scaled before modeling to reduce the effect of the right-skewed distribution.

- **Key Metrics:** The strongest visible relationship is between DrivAge and BonusMalus at approximately r = -0.48, indicating a moderate negative relationship. No numerical feature pair reaches the severe multicollinearity threshold of |r| ≥ 0.85.
